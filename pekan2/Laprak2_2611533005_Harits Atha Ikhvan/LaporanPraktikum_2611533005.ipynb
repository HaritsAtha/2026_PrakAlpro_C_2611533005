## 2.1 Dasar Teori

Python adalah salah satu bahasa pemrograman tingkat tinggi yang dikenal dengan penulisan sintaksnya yang relatif sederhana. Bahasa ini dapat dimanfaatkan untuk mengembangkan berbagai macam program, baik untuk kebutuhan dasar maupun aplikasi yang lebih kompleks. Dalam mempelajari Python, terdapat beberapa konsep utama yang perlu dipahami, di antaranya variabel, tipe data, operator, input, output, dan percabangan.

1. Variabel
Variabel digunakan untuk menyimpan sebuah nilai atau data yang nantinya dapat digunakan kembali dalam program. Pada Python, pembuatan variabel dapat dilakukan dengan langsung memberikan suatu nilai tanpa perlu mendefinisikan tipe datanya terlebih dahulu. Sebagai contoh, nilai = 85.

2. Tipe Data
Tipe data menunjukkan jenis data yang tersimpan pada sebuah variabel. Python memiliki beberapa tipe data dasar, seperti int untuk bilangan bulat, float untuk bilangan pecahan atau desimal, complex untuk bilangan kompleks, str untuk data berupa teks, serta bool untuk menyatakan kondisi benar atau salah melalui nilai True dan False.

3. Input dan Output
Input merupakan proses untuk mendapatkan data atau masukan dari pengguna. Pada Python, proses tersebut dapat dilakukan menggunakan fungsi input(). Sebaliknya, output digunakan untuk menampilkan informasi maupun hasil dari proses program dengan fungsi print(). Nilai yang diterima melalui input() pada awalnya memiliki tipe string, sehingga perlu dilakukan konversi menggunakan int() atau float() apabila data tersebut akan digunakan dalam operasi perhitungan.

4. Operator
Operator merupakan simbol yang digunakan untuk melakukan suatu operasi terhadap data. Beberapa operator aritmatika yang umum digunakan dalam Python adalah +, -, *, dan /, yang masing-masing dapat digunakan untuk melakukan perhitungan. Selain itu, terdapat operator perbandingan seperti >= yang digunakan untuk membandingkan dua nilai dan menghasilkan nilai boolean.

5. Percabangan
Percabangan berfungsi untuk menentukan perintah yang dijalankan berdasarkan kondisi tertentu. Python menyediakan pernyataan if untuk membuat percabangan. Jika terdapat lebih dari satu kondisi yang perlu diperiksa, operator logika seperti and dapat digunakan untuk menghubungkan kondisi-kondisi tersebut.

6. Konstanta
Konstanta adalah nilai yang ditetapkan untuk digunakan sebagai nilai tetap dalam suatu program. Pada Python, penggunaan Final dari modul typing dapat digunakan sebagai penanda bahwa suatu nilai ditujukan untuk tidak diubah. Contohnya adalah konstanta PI yang dapat digunakan ketika melakukan perhitungan luas lingkaran.

## 2.2 Kode Program

### 2.2.1 Program "Numerik_2611533005.py"

```python
# Buat file  dengan nama Numerik_NIM.py
# nama variable ditambah 4 digit terakhir NIM contoh: a_5500
# #create a variable with numeric type value
a_3005=100
print("The type of variable having value", a_3005, " is ", type(a_3005))

#create a variable with float value
b_3005=10.2345
print("The type of variable having value", b_3005, " is ", type(b_3005))

#create a variable with complex value.
c_3005=100+3j
print("The type of variable having value", c_3005, " is ", type(c_3005))
```

**Penjelasan :**
1. Pada bagian pertama, program mendefinisikan variabel a_3005 dengan nilai 100. Setelah itu, program menampilkan tipe data yang dimiliki variabel tersebut. Karena nilainya berupa bilangan bulat, tipe datanya adalah integer (int).

2. Selanjutnya, variabel b_3005 diisi dengan nilai 10.2345. Program kemudian mengecek dan menampilkan tipe datanya. Nilai tersebut merupakan tipe float karena mengandung bilangan desimal.

3. Pada bagian terakhir, variabel c_3005 diberi nilai 100 + 3j. Program menampilkan tipe data dari variabel tersebut, yaitu complex, karena nilai tersebut memiliki bagian imajiner yang menggunakan simbol j.


### 2.2.2 Program "String_2611533005.py"

```python
# Buat file dengan nama string_NIM.py
# nama variable ditambah 4 digit nim terakhir contoh: nama_1234
# ganti programnya dan data masing2
nama_3005 = "Harits"
jenis_kelamin_3005 = 'L'
alamat_3005 = """
    Komplek Unand
    Kec.pauh
    Kota Padang
"""
agama_3005 = 'Islam'

print("Nama saya adalah: ", nama_3005)
print("Jenis kelamin saya adalah: ", jenis_kelamin_3005)
print("Alamat saya adalah: ", alamat_3005)
print("Agama saya adalah: ", agama_3005)
```

4. Variabel agama_3005 menyimpan informasi agama dalam bentuk data teks, yaitu "Islam".
5. Selanjutnya, fungsi print() digunakan untuk menampilkan seluruh informasi yang tersimpan pada variabel nama, jenis kelamin, alamat, dan agama.


### 2.2.3 Program "Boolean_2611533005.py"

```python
# Buat file dengan nama Boolean_NIM.py
# Nama variable dtambah 4 digit terakhir contoh: nilai_1234
# Deklarasi variable dengan tipe data boolean

# # Deklarasi variabel dengan tipe data Boolean
is_lulus_3005 = True
is_cumlaude_3005 = True

# Menggunakan Boolean
nilai_3005 = 85
batas_lulus_3005 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_3005 = nilai_3005 >= batas_lulus_3005 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_3005)
print("Apakah Lulus?:", status_kelulusan_3005)
if is_lulus_3005 and is_cumlaude_3005:
    print("Selamat, Anda lulus dengan predikat Cumlaude!")
```

**Penjelasan :**

1. Program diawali dengan memberikan nilai True pada variabel is_lulus_3005 dan is_cumlaude_3005 sebagai penanda bahwa kedua kondisi tersebut bernilai benar.
2. Selanjutnya, variabel nilai_3005 diisi dengan nilai 85, sedangkan batas_lulus_3005 ditetapkan sebesar 75 sebagai batas nilai untuk dinyatakan lulus.
3. Variabel status_kelulusan_3005 diperoleh dari hasil perbandingan antara nilai_3005 dan batas_lulus_3005. Hasilnya adalah True karena nilai 85 memenuhi kondisi lebih besar atau sama dengan 75.
4. Setelah itu, fungsi print() digunakan untuk menampilkan nilai yang diperoleh serta hasil pengecekan status kelulusan.
5. Terakhir, perintah if bersama operator and digunakan untuk memastikan kedua kondisi, yaitu lulus dan Cum Laude, terpenuhi. Karena keduanya bernilai True, program menampilkan keterangan bahwa pengguna lulus dengan predikat Cum Laude.


### 2.2.4 Program "Input_2611533005.py"

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variable ditambah 4 digit nim terakhir contoh: panjang_1234
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

panjang_3005 = (input('Masukkan nilai panjang: '))
lebar_3005 = (input('Masukkan nilai lebar: '))
luas_3005 = int(panjang_3005) * int(lebar_3005)
print("Luas =", luas_3005)
```

**Penjelasan :**

1. Program meminta pengguna mengisi nilai panjang melalui input(), lalu data tersebut disimpan pada variabel panjang_3005.
2. Selanjutnya, pengguna diminta memasukkan nilai lebar yang kemudian disimpan dalam variabel lebar_3005.
3. Nilai panjang dan lebar digunakan untuk memperoleh luas persegi panjang. Karena hasil dari input() berupa teks, kedua nilai diubah menjadi bilangan bulat dengan int().
4. Setelah proses perhitungan selesai, nilai luas ditampilkan kepada pengguna melalui fungsi print()`.


### 2.2.5 Program "Konstanta_2611533005.py"

```python
# Buat nama dengan nama Konstanta_NIM.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variable ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final, final
PI: Final = 3.14
print("pi: %f" % PI)
jari_3005 = float(input('Masukkan nilai jari-jari: '))
luas_3005 = PI * jari_3005 * jari_3005
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3005, luas_3005))
```

**Penjelasan :**

1. Program diawali dengan mengambil Final dari modul typing, kemudian mendefinisikan nilai PI sebesar 3.14 sebagai konstanta.
2. Selanjutnya, program mencetak nilai PI menggunakan format angka desimal.
3. Program meminta pengguna untuk memasukkan jari-jari lingkaran. Nilai tersebut dikonversi ke tipe float dan disimpan pada variabel jari_3005.
4. Setelah mendapatkan nilai jari-jari, program menghitung luas lingkaran dengan memanfaatkan konstanta PI dan nilai jari_3005.
5. Pada bagian akhir, program menampilkan nilai jari-jari beserta hasil luas lingkaran dalam format dua angka di belakang koma.
