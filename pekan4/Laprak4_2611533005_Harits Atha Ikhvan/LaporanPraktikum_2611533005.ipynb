## 2.1 Dasar Teori

Percabangan digunakan untuk membuat program dapat menentukan tindakan berdasarkan kondisi tertentu. Pada Python, percabangan dapat digunakan untuk menghasilkan keluaran yang berbeda sesuai dengan nilai yang diberikan melalui input.

1. Percabangan if digunakan untuk menjalankan perintah jika suatu kondisi terpenuhi. Pada latihan, if digunakan untuk memeriksa kondisi umur dan IPK. Jika kondisi yang ditentukan benar, maka program akan menjalankan perintah yang ada di dalam if.

2. Percabangan if-else digunakan ketika terdapat dua kemungkinan hasil. Jika kondisi pada if terpenuhi, program menjalankan perintah pada bagian if. Jika kondisi tidak terpenuhi, program menjalankan perintah pada bagian else. Pada latihan, bentuk ini digunakan untuk menentukan apakah seseorang lulus berdasarkan IPK.

3. Percabangan if-elif-else digunakan untuk memeriksa beberapa kondisi secara berurutan. Program akan menjalankan perintah pada kondisi pertama yang terpenuhi. Pada latihan, bentuk ini digunakan untuk menentukan apakah seseorang boleh membawa motor berdasarkan umur dan kepemilikan SIM.

4. Operator Logika dan Perbandingan digunakan untuk menggabungkan kondisi dalam percabangan. Operator and digunakan ketika dua kondisi harus terpenuhi secara bersamaan. Operator != digunakan untuk memeriksa apakah suatu nilai berbeda dari nilai tertentu. Operator perbandingan seperti >= dan < digunakan untuk membandingkan nilai yang diberikan dengan suatu batas tertentu.

5. Percabangan Multi-if menggunakan beberapa perintah if yang berdiri sendiri. Setiap kondisi akan diperiksa secara terpisah sehingga beberapa kondisi dapat dijalankan jika masing-masing memenuhi syarat. Pada latihan, multi-if digunakan untuk menentukan beberapa kondisi berdasarkan umur dan kepemilikan SIM serta untuk menghitung diskon yang dapat ditambahkan.

6. Percabangan match-case digunakan untuk memilih perintah berdasarkan nilai tertentu. Setiap pilihan dibuat menggunakan case. Pada latihan, match-case digunakan untuk mengubah angka menjadi nama bulan. Terdapat juga case _ yang digunakan ketika nilai yang dimasukkan tidak sesuai dengan pilihan yang tersedia.

7. Input dan Pengolahan Data digunakan untuk menerima data dari pengguna. Data yang dimasukkan dapat diubah menjadi tipe data tertentu menggunakan int() atau float(). Pada latihan juga digunakan [0] untuk mengambil karakter pertama dari input dan strip().lower() untuk menghilangkan spasi yang tidak diperlukan serta mengubah huruf menjadi huruf kecil.

### 2.2.1 Program "if1_2611533005.py"

```python
# Buat file dengan nama if1_NIM.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

ipk_3005 = float(input("Input IPK Anda = "))

if ipk_3005 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3005))

print("Program Selesai")
```

### 2.2.2 Program "if2_2611533005.py"

```python
# Buat file dengan nama if2_NIM.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

ipk_3005 = float(input("Input IPK Anda = "))

if ipk_3005 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3005))
else:
    print("Anda Tidak Lulus")
print("Program Selesai")
```

### 2.2.3 Program "multi_if1_2611533005.py"

```python
# Buat file dengan nama multi_if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

umur_3005 = int(input("Input umur anda: "))
sim_3005 = input("Apakah Anda Sudah Punya Sim C (y/t): ")[0]

if umur_3005 >= 17 and sim_3005 == 'y' :
    print("Anda sudah dewasa dan boleh bawa motor")

if umur_3005 >= 17 and sim_3005 != 'y' :
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")

if umur_3005 < 17 and sim_3005 == 'y' :
    print("Anda belum cukup umur punya SIM")

if umur_3005 < 17 and sim_3005 != 'y' :
    print("Anda belum cukup umur bawa motor")
```

**Penjelasan :**

1. Program dimulai dengan meminta pengguna memasukkan umur dan informasi mengenai kepemilikan SIM C. Umur disimpan dalam variabel umur_3005 dan diubah menjadi tipe integer, sedangkan jawaban SIM C disimpan dalam variabel sim_3005 dengan mengambil karakter pertama dari input.

2. Setelah mendapatkan data, program menggunakan beberapa percabangan if untuk memeriksa kondisi umur dan kepemilikan SIM C. Setiap percabangan memiliki kondisi yang berbeda berdasarkan umur dan jawaban yang diberikan.

3. Jika umur pengguna sudah 17 tahun atau lebih dan memiliki SIM C, program menampilkan bahwa pengguna sudah dewasa dan boleh membawa motor. Jika tidak memiliki SIM C, program menampilkan bahwa pengguna tidak boleh membawa motor.

4. Jika umur pengguna masih di bawah 17 tahun, program akan memeriksa kembali kepemilikan SIM C. Hasil yang ditampilkan akan berbeda sesuai dengan jawaban yang diberikan, sehingga program dapat menentukan apakah pengguna cukup umur dan boleh membawa motor atau tidak.

### 2.2.4 Program "multi_if2_2611533005.py"

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_1234
# Program ini menggunakan fungsi input()
# Program menghitung diskon belanja

# Input dari user
total_belanja_3005 = float(input("Masukkan total belanja (Rp): "))

# Input status mnember (mengecek apakah user mengetik 'y' atau 'ya')
input_member__3005 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member_3005 = input_member__3005 in ["y", "ya"] 

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3005 = input("Apakah kode promo valid? (y/t) : ").strip().lower()
kode_promo_valid_3005 = input_promo_3005 in ["y", "ya"]

total_diskon_persen_3005 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus\

if total_belanja_3005 > 1000000:
    total_diskon_persen_3005 += 10 # Diskon belanja besar

if is_member_3005:
    total_diskon_persen_3005 += 5 # Diskon member

if kode_promo_valid_3005:
    total_diskon_persen_3005 += 15 # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_3005 = total_belanja_3005 * (total_diskon_persen_3005 / 100)
total_bayar_3005 = total_belanja_3005 - nominal_diskon_3005

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total DIskon : {total_diskon_persen_3005}% (Rp {nominal_diskon_3005:,.0f})")
print(f"Total Bayar  : Rp {total_bayar_3005:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_3005}%")
# Output : total diskon yang anda dapatkan: 30% jika belanja . 1 juta, member, dan kode promo valid
```

**Penjelasan :**

1. Program dimulai dengan meminta pengguna memasukkan total belanja, status member, dan status kode promo. Total belanja disimpan sebagai tipe float, sedangkan input member dan promo diubah menjadi huruf kecil agar lebih mudah diperiksa.

2. Setelah input diterima, program mengecek apakah pengguna merupakan member dan apakah kode promo yang dimasukkan valid. Hasil pengecekan tersebut disimpan dalam variabel is_member_3005 dan kode_promo_valid_3005.

3. Selanjutnya program menggunakan beberapa percabangan if yang berdiri sendiri. Setiap kondisi diperiksa secara terpisah, sehingga diskon dapat ditambahkan jika pengguna memenuhi lebih dari satu syarat. Diskon yang diberikan adalah 10% untuk belanja lebih dari Rp1.000.000, 5% untuk member, dan 15% jika kode promo valid.

4. Setelah total diskon didapatkan, program menghitung nominal diskon dari total belanja. Kemudian nominal tersebut dikurangi dari total belanja untuk mendapatkan total yang harus dibayar.

5. Terakhir, program menampilkan rincian pembayaran berupa persentase diskon, nominal diskon, dan total bayar. Program juga menampilkan kembali total diskon yang diperoleh pengguna.

### 2.2.7 Program "match_case1_2611533005.py"

```python
# Buat file dengan nama match_case1_nim.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_3005 = int(input("Masukkan angka bulan (1-12): "))

match bulan_3005:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan :**

1. Program dimulai dengan meminta pengguna memasukkan angka bulan dari 1 sampai 12. Angka tersebut disimpan ke dalam variabel bulan_3005 dan diubah menjadi tipe integer.

2. Setelah mendapatkan angka bulan, program menggunakan match-case untuk mencocokkan angka yang dimasukkan dengan pilihan case yang tersedia.

3. Setiap angka memiliki case masing-masing yang akan menampilkan nama bulan sesuai dengan angka tersebut. Contohnya, jika pengguna memasukkan angka 1, program akan menampilkan Januari.

4. Pada bagian terakhir terdapat case _ yang digunakan untuk menangani angka yang tidak sesuai dengan pilihan 1 sampai 12. Jika angka yang dimasukkan tidak sesuai, program akan menampilkan pesan "Angka tidak valid".

3. Percabangan if digunakan untuk memeriksa apakah jumlah tiket yang dimasukkan kurang dari atau sama dengan 0. Jika kondisi tersebut terpenuhi, program akan menampilkan peringatan bahwa kuota tiket tidak valid.

4. Percabangan match-case digunakan untuk menentukan harga berdasarkan paket wahana yang dipilih. Terdapat lima pilihan paket, sedangkan case _ digunakan untuk menangani pilihan paket yang tidak tersedia.

5. Percabangan if-elif-else digunakan untuk memeriksa kelayakan pengunjung pada paket Motor ATV Ekstrim. Kondisi menggunakan operator and dan != untuk memeriksa umur serta kepemilikan SIM C.

6. Percabangan if-else digunakan untuk memeriksa umur pengunjung pada paket selain Motor ATV Ekstrim. Jika umur pengunjung minimal 10 tahun, maka pengunjung dinyatakan memenuhi syarat umur.

7. Multi-if digunakan untuk menghitung diskon secara terpisah. Diskon dapat ditambahkan berdasarkan total belanja, status member, kode promo, dan jumlah tiket. Setiap kondisi diperiksa secara terpisah sehingga beberapa diskon dapat diperoleh sekaligus.

8. Setelah diskon dihitung, program menghitung nominal diskon dan total pembayaran. Kemudian if-else digunakan untuk menentukan apakah pengunjung mendapatkan Souvenir Gratis berdasarkan total pembayaran.

9. Hasil akhir ditampilkan menggunakan f-string dengan format :,.0f agar nilai harga dan pembayaran ditampilkan dengan pemisah ribuan dan tanpa angka desimal.
