## 2.1 Dasar Teori

Perulangan digunakan untuk menjalankan suatu perintah secara berulang sehingga program tidak perlu menuliskan perintah yang sama berkali-kali. Pada Python, salah satu perulangan yang digunakan adalah for. Perulangan for dapat digunakan untuk mengulang perintah berdasarkan jumlah atau urutan tertentu.

1. Perulangan for digunakan untuk menjalankan perintah secara berulang berdasarkan suatu urutan atau nilai yang ditentukan. Pada latihan, for digunakan untuk menampilkan angka dari nilai awal sampai batas tertentu dan melakukan perulangan sesuai jumlah yang diberikan melalui input.

2. Fungsi range() digunakan untuk menentukan urutan nilai yang akan diproses oleh perulangan for. range() dapat diberikan nilai awal, nilai akhir, dan langkah perulangan. Pada latihan, range() digunakan untuk membuat perulangan dari angka tertentu sampai batas yang ditentukan.

3. Perulangan for dengan kondisi digunakan ketika perulangan membutuhkan pemeriksaan kondisi tertentu. Pada latihan jumlah bilangan genap, for digunakan untuk memeriksa angka satu per satu, kemudian kondisi if digunakan untuk menentukan apakah angka tersebut merupakan bilangan genap sebelum dijumlahkan.

4. Variabel penghitung dan penjumlahan digunakan untuk menyimpan nilai selama proses perulangan berlangsung. Pada latihan, variabel digunakan sebagai penghitung perulangan dan untuk menyimpan hasil penjumlahan angka yang diproses.

5. Nested for adalah perulangan for yang berada di dalam perulangan for lainnya. Perulangan bagian dalam akan dijalankan untuk setiap satu kali perulangan bagian luar. Pada latihan, nested for digunakan untuk membuat pola menggunakan karakter, angka, dan spasi dalam beberapa baris dan kolom.

6. Parameter end pada print() digunakan untuk mengatur agar keluaran tidak langsung berpindah ke baris berikutnya. Pada latihan pola, end="" digunakan agar beberapa karakter dapat dicetak dalam satu baris. Setelah seluruh isi baris selesai dicetak, print() digunakan untuk berpindah ke baris berikutnya.

7. Perulangan naik dan turun digunakan untuk membuat urutan yang bergerak dari kecil ke besar atau dari besar ke kecil. Pada latihan, range() dengan nilai awal, batas akhir, dan langkah tertentu digunakan untuk membuat pola angka yang menurun dan menaik, seperti pada pola jam pasir kristal palindromik.


## 2.2 Kode Program

### 2.2.1 Program "perulangan_for1_2611533005.py"

```python
# Buat file dengan nama perulangan_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3005 = int(input("Masukkan jumlah perulangan : "))

for i_3005 in range(ulang_3005):
    print(f"Perulangan ke-{i_3005}")
```

**Penjelasan :**

1. Program dimulai dengan meminta pengguna memasukkan jumlah perulangan. Nilai tersebut disimpan ke dalam variabel ulang_3005 dan diubah menjadi tipe integer menggunakan int().

2. Setelah jumlah perulangan dimasukkan, program menggunakan perulangan for dengan variabel i_3005 untuk mengulang perintah sesuai nilai yang terdapat pada ulang_3005.

3. Fungsi range() digunakan untuk menentukan urutan perulangan yang dimulai dari 0 sampai satu angka sebelum nilai ulang_3005.

4. Pada setiap perulangan, program menampilkan tulisan "Perulangan ke-" diikuti dengan nilai i_3005. Nilai tersebut akan berubah setiap kali perulangan dijalankan.

5. Setelah jumlah perulangan yang ditentukan selesai, program berhenti dan seluruh hasil perulangan telah ditampilkan.

### 2.2.2 Program "perulangan_for2_2611533005.py"

```python
# Buat file dengan nama perulangan_for2_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3005 = int(input("Masukkan jumlah perulangan : "))
print("Perulangan ke-0 sampai ke-", ulang_3005-1)
for i_3005 in range(ulang_3005):
    print(i_3005, end=" ")
print()
print("Perulangan ke-1 sampai ke-", ulang_3005)
for i_3005 in range(1, ulang_3005+1):
    print(i_3005, end=" ")
```

**Penjelasan :**

1. Program dimulai dengan meminta pengguna memasukkan jumlah perulangan. Nilai tersebut disimpan dalam variabel ulang_3005 dan diubah menjadi tipe integer menggunakan int().

2. Program kemudian menampilkan keterangan bahwa perulangan pertama akan dimulai dari angka 0 sampai satu angka sebelum nilai ulang_3005.

3. Perulangan for pertama menggunakan range(ulang_3005), sehingga nilai i_3005 dimulai dari 0 dan terus bertambah sampai satu angka sebelum nilai ulang_3005. Setiap nilai ditampilkan dalam satu baris menggunakan end=" ".

4. Setelah perulangan pertama selesai, print() digunakan untuk berpindah ke baris berikutnya. Program kemudian menampilkan keterangan bahwa perulangan kedua dimulai dari angka 1 sampai nilai ulang_3005.

5. Perulangan for kedua menggunakan range(1, ulang_3005+1), sehingga nilai i_3005 dimulai dari 1 dan berakhir pada nilai ulang_3005. Setiap nilai ditampilkan menggunakan end=" " agar tetap berada dalam satu baris.

6. Dari kedua perulangan tersebut, dapat dilihat perbedaan penggunaan range(), yaitu perulangan pertama dimulai dari 0, sedangkan perulangan kedua dimulai dari 1.

### 2.2.3 Program "perulangan_for3_2611533005.py"

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3005 = int(input("Masukkan jumlah perulangan : "))

jumlah_3005 = 0
for i_3005 in range(1, ulang_3005 + 1):
    print(i_3005, end=" ")
    jumlah_3005 = jumlah_3005 + i_3005

    if i_3005 < ulang_3005:
        print(" + ", end="")
    else:
        print(" = ", jumlah_3005, end="")
print()
print("Jumlah =", jumlah_3005)
```

**Penjelasan :**

1. Program dimulai dengan meminta pengguna memasukkan jumlah perulangan. Nilai tersebut disimpan dalam variabel ulang_3005 dan diubah menjadi tipe integer menggunakan int().

2. Variabel jumlah_3005 diberi nilai awal 0. Variabel ini digunakan untuk menyimpan hasil penjumlahan selama proses perulangan berlangsung.

3. Program menggunakan perulangan for dengan variabel i_3005 dan range(1, ulang_3005 + 1) agar perulangan dimulai dari angka 1 sampai nilai yang dimasukkan pengguna.

4. Setiap angka yang sedang diproses ditampilkan menggunakan print() dengan end=" " agar angka dan tanda tambah tetap berada pada baris yang sama.

5. Pada setiap perulangan, nilai i_3005 ditambahkan ke dalam jumlah_3005 sehingga menghasilkan jumlah keseluruhan dari angka yang diproses.

6. Percabangan if digunakan untuk menentukan tanda yang ditampilkan. Jika perulangan belum mencapai angka terakhir, program menampilkan tanda "+". Jika sudah mencapai angka terakhir, program menampilkan tanda "=" diikuti dengan hasil penjumlahannya.

7. Setelah perulangan selesai, print() digunakan untuk berpindah ke baris berikutnya, kemudian program menampilkan hasil akhir dengan tulisan "Jumlah =" beserta nilai jumlah_3005.

### 2.2.4 Program "nested_for1_2611533005.py"

```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_3005 = int(input("Masukkan nilai batas: "))
for line_3005 in range(1, batas_3005 + 1):
    for j_3005 in range(1, (-1 * line_3005 + batas_3005) + 1):
        print(".", end=" ")
    print(line_3005)
```

**Penjelasan :**

1. Program dimulai dengan meminta pengguna memasukkan nilai batas. Nilai tersebut disimpan ke dalam variabel batas_3005 dan diubah menjadi tipe integer menggunakan int().

2. Setelah nilai batas dimasukkan, program menggunakan perulangan for dengan variabel line_3005 yang dimulai dari 1 sampai nilai batas_3005.

3. Di dalam perulangan tersebut terdapat nested for dengan variabel j_3005. Perulangan ini digunakan untuk menentukan jumlah titik yang akan ditampilkan pada setiap baris.

4. Jumlah perulangan pada j_3005 ditentukan menggunakan rumus (-1 * line_3005 + batas_3005) + 1. Dengan rumus tersebut, jumlah titik akan berkurang setiap kali line_3005 bertambah.

5. Setiap titik ditampilkan menggunakan print() dengan end=" " agar titik-titik tetap berada pada baris yang sama dan memiliki jarak satu spasi.

6. Setelah perulangan j_3005 selesai, print(line_3005) digunakan untuk berpindah ke baris berikutnya sekaligus menampilkan nomor baris yang sedang diproses.

### 2.2.5 Program "nested_for2_2611533005.py"

```python
# Buat file dengan nama nested_for2_NIM.py
# Buat program untuk perulangan dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_3005 = int(input("Masukkan nilai batas: "))
for i_3005 in range(1, batas_3005 + 1):
    for j_3005 in range(1, batas_3005 + 1):
        print("*", end=" ")
    print() # pindah ke baris berikutnya
```

**Penjelasan :**

2. Program kemudian menggunakan perulangan for pertama dengan variabel i_3005 untuk menentukan jumlah baris yang akan dibuat, dimulai dari 1 sampai nilai batas_3005.

3. Di dalam perulangan pertama terdapat perulangan for kedua dengan variabel j_3005. Perulangan ini digunakan untuk menentukan jumlah karakter * yang dicetak pada setiap baris.

4. Perintah print() menggunakan end=" " agar setiap karakter * tetap berada pada baris yang sama dan dipisahkan dengan spasi.

5. Setelah perulangan kedua selesai, print() tanpa argumen digunakan untuk berpindah ke baris berikutnya. Proses ini terus dilakukan sampai jumlah baris sesuai dengan nilai batas_3005.

### 2.2.6 Program "nested_for3_2611533005.py"

```python
# Buat file dengan nama nested_for3_NIM.py
# Buat program untuk perulangan dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_3005 = int(input("Masukkan nilai batas: "))
for i_3005 in range(batas_3005 + 1):
    for j_3005 in range(batas_3005+1):
        print(i_3005+j_3005, end=" ")
    print() # pindah ke baris berikutnya
```

3. Di dalam perulangan pertama terdapat perulangan for kedua dengan variabel j_3005. Perulangan ini digunakan untuk menentukan nilai pada setiap posisi di dalam baris dan juga berjalan dari 0 sampai nilai batas_3005.

4. Pada setiap perulangan, program menghitung hasil dari i_3005 + j_3005 kemudian menampilkannya menggunakan print(). Parameter end=" " digunakan agar nilai berikutnya tetap dicetak pada baris yang sama.

5. Setelah perulangan kedua selesai, perintah print() digunakan untuk berpindah ke baris berikutnya. Proses ini terus dilakukan sampai seluruh baris sesuai batas yang diberikan selesai ditampilkan.

### 2.2.7 Program "nested_for4_2611533005.py"

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangna dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_3005 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3005 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3005 = tinggi_3005
    c_3005 = a_3005
    lebar_3005 = (2*tinggi_3005) - 2

    for i_3005 in range(1, tinggi_3005 + 1):
        b_3005 = c_3005 + 1

        for j_3005 in range(1, lebar_3005 + 1):

            # Baris atas dan bawah
            if i_3005 == 1 or i_3005 == tinggi_3005:
                if j_3005 == 1 or j_3005 == lebar_3005:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_3005 == 1 or j_3005 == lebar_3005:
                    print("|", end="")
                else:
                    if j_3005 == c_3005:
                        print("<", end="")
                    elif j_3005 == b_3005:
                        print(">", end="")
                    elif j_3005 == (lebar_3005 - c_3005):
                        print("<", end="")
                    elif j_3005 == (lebar_3005 - c_3005 + 1):
                        print(">", end="")
                    elif j_3005 > b_3005 and j_3005 < (lebar_3005 - c_3005):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_3005 -= 2

        if a_3005 <= 0:
            c_3005 = (-a_3005) + 2
        else:
            c_3005 = a_3005
```

**Penjelasan :**

1. Program dimulai dengan meminta pengguna memasukkan tinggi pola. Nilai tersebut disimpan ke dalam variabel tinggi_3005 dan diubah menjadi tipe integer menggunakan int().

2. Program kemudian memeriksa apakah nilai tinggi_3005 merupakan bilangan genap menggunakan operator %. Jika nilai tersebut merupakan bilangan ganjil, program akan menampilkan pesan bahwa tinggi harus berupa bilangan genap.

3. Jika nilai yang dimasukkan merupakan bilangan genap, program akan menentukan beberapa nilai awal untuk membentuk pola. Variabel a_3005 dan c_3005 digunakan untuk mengatur posisi pola, sedangkan lebar_3005 digunakan untuk menentukan lebar keseluruhan pola.

4. Perulangan for pertama menggunakan variabel i_3005 untuk membuat baris pola dari baris pertama sampai baris terakhir. Di dalamnya terdapat perulangan for kedua menggunakan j_3005 untuk mencetak setiap bagian pola dari kiri ke kanan.

5. Pada baris pertama dan terakhir, program menggunakan karakter # pada bagian ujung dan = pada bagian tengah sehingga membentuk bingkai atas dan bawah.

6. Pada baris lainnya, program menggunakan karakter | sebagai batas kiri dan kanan. Beberapa kondisi if dan elif digunakan untuk menentukan posisi karakter <, >, ., dan spasi sehingga membentuk pola di bagian dalam.

7. Setelah satu baris selesai dicetak, nilai a_3005 dikurangi 2 untuk mengubah posisi pola pada baris berikutnya. Nilai c_3005 kemudian diperbarui berdasarkan hasil perhitungan tersebut sehingga posisi karakter pada pola dapat bergerak sesuai bentuk yang diinginkan.

8. Setelah seluruh baris selesai diproses oleh perulangan, program berhenti dan pola yang terbentuk ditampilkan secara lengkap.

### 2.2.8 Program "jumlah_genap_2611533005.py"

```python
# Buat file dengan nama jumlah_genap_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3005 = int(input("Masukkan nilai batas: "))

jumlah_3005 = 0
for i_3005 in range(1, ulang_3005 + 1):
    if i_3005 % 2 == 0:
        print(i_3005, end=" ")
        jumlah_3005 = jumlah_3005 + i_3005

        if i_3005 < ulang_3005:
            print(" + ", end="")
        else:
            print(" = ", jumlah_3005, end="")
print()
print("Jumlah =", jumlah_3005)
```

**Penjelasan :**

1. Program dimulai dengan meminta pengguna memasukkan nilai batas. Nilai tersebut disimpan ke dalam variabel ulang_3005 dan diubah menjadi tipe integer menggunakan int().

2. Variabel jumlah_3005 diberi nilai awal 0. Variabel ini digunakan untuk menyimpan hasil penjumlahan bilangan genap selama proses perulangan.

3. Program menggunakan perulangan for dengan variabel i_3005 untuk memeriksa angka dari 1 sampai nilai batas yang dimasukkan. Fungsi range() digunakan untuk menentukan urutan angka tersebut.

4. Di dalam perulangan, program menggunakan kondisi if dengan operator % untuk memeriksa apakah i_3005 merupakan bilangan genap. Jika sisa pembagian angka dengan 2 adalah 0, maka angka tersebut merupakan bilangan genap.

5. Bilangan genap yang ditemukan ditampilkan menggunakan print() dengan end="" agar angka dan tanda tambah tetap berada pada baris yang sama. Nilai tersebut kemudian ditambahkan ke dalam jumlah_3005.

6. Program kembali menggunakan kondisi if untuk memeriksa apakah angka yang sedang diproses masih lebih kecil dari nilai batas. Jika masih ada angka berikutnya, program menampilkan tanda "+". Jika sudah mencapai batas, program menampilkan tanda "=" diikuti dengan hasil penjumlahan.

7. Setelah seluruh perulangan selesai, print() digunakan untuk berpindah ke baris berikutnya, kemudian program menampilkan hasil akhir penjumlahan bilangan genap yang tersimpan dalam jumlah_3005.

```python
n_3005 = int(input("Masukkan ukuran skala jam pasir (N): "))

lebar_3005 = 4 * n_3005 + 7

for j_3005 in range(lebar_3005):
    if j_3005 == 0 or j_3005 == lebar_3005 - 1:
        print("#", end="")
    else:
        print("=", end="")
print()

for baris_3005 in range(n_3005, 0, -1):
    print("|", end="")

    spasi_3005 = 2 * (n_3005 - baris_3005) + 1
    for j_3005 in range(spasi_3005):
        print(" ", end="")

    for angka_3005 in range(baris_3005, 0, -1):
        print(angka_3005, end=" ")

    print("<*>", end="")

    for angka_3005 in range(1, baris_3005 + 1):
        print(" ", end="")
        print(angka_3005, end="")

    for j_3005 in range(spasi_3005):
        print(" ", end="")

    print("|")

print("|", end="")

spasi_3005 = 2 * n_3005 + 1
for j_3005 in range(spasi_3005):
    print(" ", end="")

print("<*>", end="")

for j_3005 in range(spasi_3005):
    print(" ", end="")

print("|")

for baris_3005 in range(1, n_3005 + 1):
    print("|", end="")

    spasi_3005 = 2 * (n_3005 - baris_3005) + 1
    for j_3005 in range(spasi_3005):
        print(" ", end="")

    for angka_3005 in range(baris_3005, 0, -1):
        print(angka_3005, end=" ")

    print("<*>", end="")

    for angka_3005 in range(1, baris_3005 + 1):
        print(" ", end="")
        print(angka_3005, end="")

    for j_3005 in range(spasi_3005):
        print(" ", end="")

    print("|")

for j_3005 in range(lebar_3005):
    if j_3005 == 0 or j_3005 == lebar_3005 - 1:
        print("#", end="")
    else:
        print("=", end="")
print()
```
