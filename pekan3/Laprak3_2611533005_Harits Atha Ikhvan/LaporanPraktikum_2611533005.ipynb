## 2.1 Dasar Teori

Operator merupakan simbol yang digunakan untuk melakukan suatu operasi terhadap data dalam program. Dalam Python, terdapat beberapa jenis operator yang memiliki fungsi berbeda, mulai dari melakukan perhitungan sampai membandingkan dan mengolah nilai.

1. Operator Aritmatika
   Operator aritmatika digunakan untuk melakukan perhitungan. Beberapa operator yang digunakan dalam Python yaitu + untuk penjumlahan, - untuk pengurangan, * untuk perkalian, / untuk pembagian, // untuk pembagian bulat, % untuk sisa bagi, dan ** untuk perpangkatan.

2. Operator Assignment
   Operator assignment digunakan untuk memberikan atau mengubah nilai sebuah variabel. Selain =, terdapat juga operator seperti +=, -=, *=, /=, //=, %=, dan **= yang dapat digunakan untuk melakukan operasi sekaligus mengubah nilai variabel.

3. Operator Perbandingan
   Operator perbandingan digunakan untuk membandingkan dua nilai dan menghasilkan nilai True atau False. Operator yang digunakan antara lain >, <, >=, <=, ==, dan !=.

4. Operator Logika
   Operator logika digunakan untuk menggabungkan atau membalik suatu kondisi. Python memiliki operator and, or, dan not. Operator and menghasilkan True jika kedua kondisi benar, sedangkan or menghasilkan True jika salah satu kondisi benar. Operator not digunakan untuk membalik nilai kondisi.

5. Operator Keanggotaan
   Operator keanggotaan digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah data seperti list. Operator in digunakan untuk mengecek apakah nilai terdapat di dalam data, sedangkan not in digunakan untuk mengecek apakah nilai tidak terdapat di dalam data.

6. Operator Identitas
   Operator identitas digunakan untuk membandingkan apakah dua variabel merujuk pada objek yang sama. Operator is digunakan untuk memeriksa identitas objek, sedangkan is not digunakan untuk memeriksa apakah identitas kedua objek berbeda. Operator ini berbeda dengan == yang membandingkan nilai dari objek.

7. Operator Bitwise
   Operator bitwise digunakan untuk melakukan operasi pada bilangan dalam bentuk bit atau biner. Beberapa operator yang digunakan yaitu & untuk AND, | untuk OR, ^ untuk XOR, ~ untuk NOT, << untuk geser kiri, dan >> untuk geser kanan.


## 2.2 Kode Program

### 2.2.1 Program "aritmatika_2611533005.py"

```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_3005 = int(input("Input angka-1: "))
angka2_3005 = int(input("Input angka-2: "))

# Penjumlahan
hasil_3005 = angka1_3005 + angka2_3005
print("\nOperator Penjumlahan")
print("Hasil =", hasil_3005)

# Pengurangan
hasil_3005 = angka1_3005 - angka2_3005
print("\nOperator Pengurangan")
print("Hasil =", hasil_3005)

# Perkalian
hasil_3005 = angka1_3005 * angka2_3005
print("\nOperator Perkalian")
print("Hasil =", hasil_3005)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_3005 != 0:
    hasil_3005 = angka1_3005 / angka2_3005
    print("\nOperator Pembagian")
    print("Hasil =", hasil_3005)

    hasil_3005 = angka1_3005 // angka2_3005
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_3005)

    hasil_3005 = angka1_3005 % angka2_3005
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_3005)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_3005 = angka1_3005 ** angka2_3005
print("\nOperator Pangkat")
print("Hasil =", hasil_3005)
```

**Penjelasan :**

1. Program dimulai dengan meminta pengguna memasukkan dua angka. Kedua angka tersebut disimpan ke dalam variabel angka1_3005 dan angka2_3005, kemudian diubah menjadi tipe integer agar bisa digunakan dalam perhitungan.

2. Setelah mendapatkan kedua angka, program melakukan beberapa operasi dasar, yaitu penjumlahan, pengurangan, dan perkalian. Masing-masing operasi menggunakan operator yang berbeda dan hasilnya ditampilkan setelah perhitungan selesai.

3. Selanjutnya program melakukan pembagian, pembagian bulat, dan sisa bagi. Sebelum melakukan operasi tersebut, program mengecek apakah angka kedua tidak sama dengan 0 agar pembagian dapat dilakukan dengan benar.

4. Pada bagian terakhir, program menggunakan operator pangkat untuk menghitung angka pertama yang dipangkatkan dengan angka kedua. Hasil dari setiap operasi disimpan dalam variabel hasil_3005 sebelum ditampilkan.

### 2.2.2 Program "assignment_2611533005.py"

```python
# Buat file dengan nama assigment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assigment dalam Python

angka1_3005 = int(input("Input angka-1: "))
angka2_3005 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =",angka1_3005)
print("Nilai angka2 =",angka2_3005)

# Assigment biasa
hasil_3005 = angka1_3005 
print("\nAssigment biasa (=)")
print("Hasil =",hasil_3005)

# Assigment penambahan
hasil_3005 = angka1_3005 
hasil_3005 += angka2_3005 
print("\nAssigment penambahan (+=)")
print("Hasil =",hasil_3005)

# Assigment pengurangan
hasil_3005 = angka1_3005 
hasil_3005 -= angka2_3005 
print("\nAssigment pengurangan (-=)")
print("Hasil =",hasil_3005)

# Assigment perkalian
hasil_3005 = angka1_3005 
hasil_3005 *= angka2_3005
print("\nAssigment perkalian (*=)")
print("Hasil =",hasil_3005)

# Assigment pembagian, pembagian bulat, dan sisa bagi
if angka2_3005 != 0:
    hasil_3005 = angka1_3005 
    hasil_3005 /= angka2_3005
    print("\nAssigment pembagian (/=)")
    print("Hasil =",hasil_3005)
    # Operator tambahan
    hasil_3005 = angka1_3005 
    hasil_3005 //= angka2_3005
    print("\nAssigment pembagian bulat (//=)")
    print("Hasil =",hasil_3005)
    hasil_3005 = angka1_3005 
    hasil_3005 %= angka2_3005
    print("\nOperator sisa bagi (%=)")
    print("Hasil =",hasil_3005)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assigment perpangkatan
hasil_3005 = angka1_3005 
hasil_3005 **= angka2_3005
print("\nOperator perpangkatan (**=)")
print("Hasil =",hasil_3005)
```

**Penjelasan :**

1. Program diawali dengan meminta pengguna memasukkan dua angka yang disimpan dalam variabel angka1_3005 dan angka2_3005. Kedua nilai tersebut diubah menjadi tipe integer menggunakan int().

2. Setelah nilai awal ditampilkan, program mencoba beberapa bentuk assignment, yaitu assignment biasa, penambahan, pengurangan, dan perkalian. Operasi tersebut menggunakan tanda =, +=, -=, dan *= untuk mengubah nilai pada variabel hasil_3005.

3. Program kemudian melakukan assignment pembagian, pembagian bulat, dan sisa bagi menggunakan /=, //=, dan %=. Sebelum itu, program mengecek apakah angka kedua bernilai 0. Jika bernilai 0, pembagian tidak dilakukan agar tidak terjadi kesalahan.

4. Pada bagian terakhir, program menggunakan operator **= untuk melakukan perpangkatan. Setiap operasi menghasilkan nilai baru pada hasil_3005 yang kemudian ditampilkan ke layar.

### 2.2.3 Program "bitwise_2611533005.py"

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("======================================")
print("3. OPERATOR BITWISE")
print("======================================")

angka1_3005 = int(input("Masukkan angka bitwise-1: "))
angka2_3005 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =",angka1_3005,"| biner =",bin(angka1_3005))
print("angka1 =",angka2_3005,"| biner =",bin(angka2_3005))

# Bitwise AND
hasil_3005 = angka1_3005 & angka2_3005
print("\nBitwise AND (&)")
print(angka1_3005,"&",angka2_3005,hasil_3005)
print("Biner hasil =",bin(hasil_3005))
print("Biner hasil (8 bit) =",format(hasil_3005,"08b"))

# Bitwise OR
hasil_3005 = angka1_3005 | angka2_3005
print("\nBitwise OR (|)")
print(angka1_3005,"|",angka2_3005,hasil_3005)
print("Biner hasil =",bin(hasil_3005))
print("Biner hasil (8 bit) =",format(hasil_3005,"08b"))

# Bitwise XOR
hasil_3005 = angka1_3005 ^ angka2_3005
print("\nBitwise XOR (^)")
print(angka1_3005,"^",angka2_3005,hasil_3005)
print("Biner hasil =",bin(hasil_3005))
print("Biner hasil (8 bit) =",format(hasil_3005,"08b"))

# Bitwise NOT
hasil_3005 = ~angka1_3005
print("\nBitwise NOT (~)")
print(angka1_3005,"~",angka2_3005,hasil_3005)
print("Biner hasil =",bin(hasil_3005))
print("Biner hasil (8 bit) =",format(hasil_3005,"08b"))

# Bitwise geser kiri
jumlah_geser_3005 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_3005 = angka1_3005 << jumlah_geser_3005
print("\nBitwise geser kiri (<<)")
print(angka1_3005,"<<",jumlah_geser_3005,"=",hasil_3005)
print("Biner hasil =",bin(hasil_3005))
print("Biner hasil (8 bit) =",format(hasil_3005,"08b"))

# Bitwise geser kanan
hasil_3005 = angka1_3005 >> jumlah_geser_3005
print("\nBitwise geser kanan (>>)")
print(angka1_3005,">>",jumlah_geser_3005,"=",hasil_3005)
print("Biner hasil =",bin(hasil_3005))
print("Biner hasil (8 bit) =",format(hasil_3005,"08b"))
```

**Penjelasan :**

1. Program meminta pengguna memasukkan dua angka, yaitu angka1_3005 dan angka2_3005. Kedua angka tersebut diubah menjadi integer dan ditampilkan dalam bentuk desimal serta biner menggunakan fungsi bin().

2. Program melakukan operasi Bitwise AND, OR, dan XOR menggunakan operator &, |, dan ^. Setiap hasil operasi ditampilkan dalam bentuk desimal dan biner agar perbedaan hasil dari setiap operator dapat dilihat.

3. Selanjutnya, program menggunakan operator Bitwise NOT (~) untuk membalik bit dari angka pertama. Hasilnya kemudian ditampilkan dalam bentuk desimal dan biner.

4. Pada bagian terakhir, pengguna memasukkan jumlah pergeseran bit. Angka pertama kemudian digeser ke kiri menggunakan << dan ke kanan menggunakan >>. Hasil kedua operasi tersebut ditampilkan dalam bentuk desimal dan biner 8 bit.

### 2.2.4 Program "lainnya_2611533005.py"

```python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("======================================")
print("1. OPERATOR KEANGGOTAAN")
print("======================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_3005 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_3005 = [int(angka.strip()) for angka in input_data_3005.split(",")]

nilai_dicari_3005 = int(input("Masukkan nilai yang ingin dicari: "))

# Operator in
hasil_3005 = nilai_dicari_3005 in data_3005
print("\nOperator keanggotaan IN")
print(nilai_dicari_3005, "in", data_3005, "=", hasil_3005)

# Operator not in
hasil_3005 = nilai_dicari_3005 not in data_3005
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_3005, "not in", data_3005, "=", hasil_3005)


print("======================================")
print("2. OPERATOR IDENTIAS")
print("======================================")

# objek1 menggunakan list dari input perngguna
object1_3005 = data_3005

# objek2 merujuk pada objek yang sama dengan objek1
object2_3005 = object1_3005

# objek3 memiliki isi sama, tetapi merupakan objek baru
object3_3005 = data_3005.copy()

print("object1 =",object1_3005)
print("object2 =",object2_3005)
print("object3 =",object3_3005)

# Operator is
hasil_3005 = object1_3005 is object2_3005
print("\nOperator identitas IS")
print("objek1 is objek2 =",hasil_3005)

# Operator is not
hasil_3005 = object1_3005 is not object3_3005
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =",hasil_3005)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =",object1_3005 is object3_3005)
print("objek1 == objek3 =",object1_3005 == object3_3005)
```

**Penjelasan :**

1. Program meminta pengguna memasukkan beberapa angka yang dipisahkan dengan koma. Data tersebut kemudian diubah menjadi list dan disimpan pada variabel data_3005.

2. Selanjutnya, pengguna memasukkan nilai yang ingin dicari. Operator in digunakan untuk mengecek apakah nilai tersebut ada di dalam list, sedangkan not in mengecek apakah nilai tersebut tidak ada.

3. Pada bagian identitas, object1_3005 dan object2_3005 mengarah ke objek list yang sama. Sementara itu, object3_3005 dibuat sebagai objek baru dengan isi yang sama menggunakan copy().

4. Operator is digunakan untuk membandingkan apakah dua objek memiliki identitas yang sama, sedangkan is not mengecek apakah identitasnya berbeda. Terakhir, operator == digunakan untuk membandingkan isi atau nilai kedua objek.

### 2.2.5 Program "logika_2611533005.py"

```python
# Buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# program ini menggunakan fungsi input()
# Program operator logika dalam Python

# Memasukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_3005 = input("Input nilai boolean-1 (True/False): ").strip().lower() == "true"
a2_3005 = input("Input nilai boolean-2 (True/False): ").strip().lower() == "true"

print("\nA1 = ", a1_3005)
print("A2 = ", a2_3005)

# Konjungsi: bernilai True jika keduanya True
hasil_3005 = a1_3005 and a2_3005
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil_3005)

# Disjungsi: bernilai True jika salah satunya True
hasil_3005 = a1_3005 or a2_3005
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil_3005)

# Negasi A1: membalik nilai A1
hasil_3005 = not a1_3005
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil_3005)

# Negasi A2: membalik nilai A2
hasil_3005 = not a2_3005
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil_3005)

# XOR: bernilai True jika kedua nilai berbeda  
hasil_3005 = a1_3005 != a2_3005
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil_3005)
```

**Penjelasan :**

1. Program meminta dua nilai boolean dari pengguna, yaitu True atau False. Input kemudian diubah menjadi nilai boolean dan disimpan pada variabel a1_3005 dan a2_3005.

2. Selanjutnya, program menggunakan operator and untuk mengecek apakah kedua nilai bernilai True, serta operator or untuk mengecek apakah salah satu nilainya True.

3. Operator not digunakan untuk membalik nilai boolean. Jika nilai awal True, hasilnya menjadi False, begitu juga sebaliknya.

4. Pada bagian terakhir, program menggunakan operator != sebagai XOR untuk mengecek apakah kedua nilai memiliki kondisi yang berbeda. Hasil dari setiap operasi ditampilkan dalam bentuk True atau False.

### 2.2.6 Program "perbandingan_2611533005.py"

```python
# Buat file dengan nama perbandingan_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator pebandingan dalam Python

angka1_3005 = int(input("Input angka-1: "))
angka2_3005 = int(input("Input angka-2: "))

# Lebih besar dari
hasil_3005 = angka1_3005 > angka2_3005
print("\nOperator lebih besar dari")
print("angka1_3005 > angka2_3005 = ", hasil_3005)

# Lebih kecil dari
hasil_3005 = angka1_3005 < angka2_3005
print("\nOperator lebih kecil dari")
print("angka1_3005 < angka2_3005 =", hasil_3005)

# Lebih besar dari atau sama dengan
hasil_3005 = angka1_3005 >= angka2_3005
print("\nOperator lebih besar dari atau sama dengan")
print("angka1_3005 >= angka2_3005 =", hasil_3005)

# Lebih kecil dari atau sama dengan
hasil_3005 = angka1_3005 <= angka2_3005
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1_3005 <= angka2_3005 =", hasil_3005)

# Sama dengan
hasil_3005 = angka1_3005 == angka2_3005
print("\nOperator sama dengan")
print("angka1_3005 == angka2_3005 =", hasil_3005)

# Tidak sama dengan
hasil_3005 = angka1_3005 != angka2_3005
print("\nOperator tidak sama dengan")
print("angka1_3005 != angka2_3005 =", hasil_3005)

# Tambahan: perbandingan berantai dalam Python
hasil = 0 < angka1_3005 < 100
print("\nPerbandingan berantai")
print("0 < angka1_3005 < 100 =", hasil)

hasil = 0 < angka2_3005 < 100
print("0 < angka2_3005 < 100 =", hasil)
```

**Penjelasan :**

1. Program meminta pengguna memasukkan dua angka, yaitu angka1_3005 dan angka2_3005. Kedua input diubah menjadi integer agar dapat digunakan dalam perbandingan.

2. Program kemudian membandingkan kedua angka menggunakan operator >, <, >=, <=, ==, dan !=. Setiap perbandingan menghasilkan nilai True atau False sesuai dengan kondisi kedua angka.

3. Selain itu, program menggunakan perbandingan berantai untuk mengecek apakah angka1_3005 dan angka2_3005 berada di antara 0 sampai 100.

4. Semua hasil perbandingan disimpan dalam variabel hasil_3005 atau hasil, lalu ditampilkan ke layar agar pengguna dapat melihat hasil dari setiap operator.
